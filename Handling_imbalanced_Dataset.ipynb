Imbalanced dataset:
A dataset with an unequal class distribution


```python
#importing the dependencies
import numpy as np
import pandas as pd
```

```python
#loading the dataset to pandas df
credit_card_data = pd.read_csv('/credit_data.csv')
```

```python
credit_card_data.head()
```

```python
credit_card_data.tail()
```

```python
#distribution of the two classes
credit_card_data['Class'].value_counts()
```

0 --> correct transaction

1--> fraud

highly imbalanced dataset

```python
#separating the legit and fraudalent trasactions
legit=credit_card_data[credit_card_data.Class==0]
fraud=credit_card_data[credit_card_data.Class==1]
```

```python
print(legit.shape)
print(fraud.shape)
```

Under-sampling technique

Building a sample dataset containing similar distribution of Legit and Fraud trasaction

Number of Fraudlent transactions -->492

```python
legit_sample = legit.sample(n=183, random_state=2)
```

```python
legit_sample.shape
```

Concatenate Two Dataframes

```python
new_dataset=pd.concat([legit_sample,fraud],axis=0)
```

```python
new_dataset.head()
```

```python
new_dataset.tail()
```

```python
new_dataset['Class'].value_counts()
```
