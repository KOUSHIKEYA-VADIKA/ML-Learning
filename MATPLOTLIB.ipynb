Matplotlib:

*   Useful for making plots

```python
#importing matplotlib library
import matplotlib.pyplot as plt
```

```python
#import numpy to get data for plots
import numpy as np
```

```python
x=np.linspace(0,10,100)
y=np.sin(x)
z=np.cos(x)
```

```python
print(x)
```

```python
print(y)
```

```python
print(z)
```

```python
#sine wave
plt.plot(x,y)
plt.show()
```

```python
#cosine wave
plt.plot(x,z)
plt.show()
```

```python
#adding title x-axis and y-axis labels
plt.plot(x,y)
plt.xlabel('angle')
plt.ylabel('sine')
plt.title('sine wave')
plt.show()
```

```python
#parabola
x=np.linspace(-10,10,20)
y=x**2
plt.plot(x,y)
plt.xlabel('x')
plt.ylabel('y=x^2')
plt.title('PARABOLA')
plt.show()
```

```python
plt.plot(x,y,'m+')
plt.show()
```

```python
plt.plot(x, y, 'g^')
plt.show()
```

```python
plt.plot(x,y,'rx')
```

```python
x=np.linspace(-5,5,50)
plt.plot(x,np.sin(x),'g-')
plt.plot(x,np.cos(x),'r--')
plt.show()
```

BAR PLOT

```python
fig=plt.figure()
ax=fig.add_axes([0,0,1,1]) #two zeros are origins and 1,1 are width and height
langs=['english','french','spanish','latin','german']
people=[100,50,150,40,70]
ax.bar(langs,people)
ax.set_xlabel('languages')
ax.set_ylabel('number of speakers')
ax.set_title('speakers in different languages')
plt.show()
```

```python
fig1=plt.figure()
ax=fig1.add_axes([0,0,1,1])
langs=['english','french','spanish','latin','german']
people=[100,50,150,40,70]
ax.pie(people,labels=langs,autopct='%1.1f%%')
plt.show()
```

Scatter Plot

```python
x=np.linspace(0,10,30)
y=np.sin(x)
z=np.cos(x)
fig2=plt.figure()
ax=fig2.add_axes([0,0,1,1])
ax.scatter(x,y,color='g')
ax.scatter(x,z,color='b')
plt.show()
```

3D Scatter Plot

```python
fig3=plt.figure()
ax=plt.axes(projection='3d')
z=20*np.random.random(100)
x=np.sin(z)
y=np.cos(z)
ax.scatter(x,y,z,c=z,cmap='Greens')
plt.show()
```
