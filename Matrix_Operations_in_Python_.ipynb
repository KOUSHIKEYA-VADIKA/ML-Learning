```python
import numpy as np
```

Matrix addition

```python
m1=np.array([[2,3],[4,5],[5,6]])
m2=np.array([[7,3],[98,5],[5,76]])
print(m1,"\n")
print(m2,"\n")
print(m1+m2)


print(np.add(m1,m2))
```

Matrix subtraction

```python
m1=np.array([[2,3],[4,5],[5,6]])
m2=np.array([[7,3],[98,5],[5,76]])
print(m1,"\n")
print(m2,"\n")
print(m1-m2)

print(np.subtract(m1,m2))
```

matrix Multiplication with Scaler

```python
x=5
y=np.random.randint(10,size=(x,x))
print(y)
product=np.multiply(x,y)
print(product)
```
