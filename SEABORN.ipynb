```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

note : Seaborn has some built in datasets

```python
#total bill vs tip dataset
tips = sns.load_dataset('tips')
```

```python
tips.head()
```

```python
#setting theme for the plots
sns.set_theme()
```

```python
#visualize the data
sns.relplot(data=tips,x='total_bill',y='tip',col='time',hue='smoker',style='smoker')
```

```python
from sklearn.datasets import fetch_california_housing
```

```python
house_california=fetch_california_housing()
```

```python
house=pd.DataFrame(house_california.data,columns=house_california.feature_names)
house['target']=house_california.target
```

```python
house.head()
```

Distribution plot

```python
sns.displot(house['HouseAge'])
```

```python
sns.distplot(house['HouseAge'])
```

Correlation matrix :
1.positive and 2.negative

Heat Map

```python
correlation=house.corr()
```

```python
#constructing a Heat Map
plt.figure(figsize=(10,10))
sns.heatmap(correlation,cbar=True,square=True,fmt='.1f',annot=True,annot_kws={'size':8},cmap='Blues')
```
