```python
import numpy as np
```

Dot Product of 2 Vectors

```python
a=np.array([2,3])
b=np.array([4,4])
a_dot_b=np.dot(a,b)
print(a_dot_b)
```

```python
c=np.array([40,20,35])
d=np.array([53,24,68])
c_dot_d=np.dot(c,d)
print(c_dot_d)
```

Cross Product

```python
a=np.array([2,3,6])
b=np.array([4,4,5])
a_cross_b=np.cross(a,b)
print(a_cross_b)
```

Projection of "a" vector on "v" vector

```python
a=np.array([2,3,6])
v=np.array([4,4,5])
#magnitute of "v" vector
magnitude_of_v=np.sqrt(sum(v**2))
#formulae
proj_of_a_on_v=(np.dot(a,v)/magnitude_of_v**2)*v
print(proj_of_a_on_v)
```
